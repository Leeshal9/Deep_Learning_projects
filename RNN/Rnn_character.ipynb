```python
import tensorflow as tf
import numpy as np

print("TensorFlow version:", tf.__version__)
```

Step 2 — Download the Shakespeare text

```python
path_to_file = tf.keras.utils.get_file(
    "shakespeare.txt",
    "https://storage.googleapis.com/download.tensorflow.org/data/shakespeare.txt"
)

print(path_to_file)
```

Step 3 — Read the Shakespeare text

```python
text = open(path_to_file, "rb").read().decode(encoding="utf-8")

print("Length of text:", len(text))
print()
print(text[:500])
```

Step 4 — Find all unique characters

```python
vocab = sorted(set(text))

print("Number of unique characters:", len(vocab))
print(vocab)
```

Step 5 — Create character → number mapping

```python
char_to_int = {char: i for i, char in enumerate(vocab)}

print(char_to_int)
```

Step 6 — Convert the entire text into numbers

```python
text_as_int = np.array([char_to_int[char] for char in text])

print("Original text:")
print(text[:100])

print("\nNumerical representation:")
print(text_as_int[:100])
```

Step 7 — Set sequence length

```python
seq_length = 100

print("Sequence length:", seq_length)
```

Step 8 — Create sequences

```python
char_dataset = tf.data.Dataset.from_tensor_slices(text_as_int)

sequences = char_dataset.batch(seq_length + 1, drop_remainder=True)

print("Number of sequences:", len(list(sequences)))
```

Step 9 — Separate input and target

```python
def split_input_target(chunk):
    input_text = chunk[:-1]
    target_text = chunk[1:]

    return input_text, target_text

dataset = sequences.map(split_input_target)

for input_example, target_example in dataset.take(1):
    print("Input:")
    print(input_example.numpy())

    print("\nTarget:")
    print(target_example.numpy())
```

Step 10 — Shuffle and batch the sequences

```python
BATCH_SIZE = 64
BUFFER_SIZE = 10000

dataset = (
    dataset
    .shuffle(BUFFER_SIZE)
    .batch(BATCH_SIZE, drop_remainder=True)
)

print("Batch size:", BATCH_SIZE)
```

Step 11 — Define the model

```python
vocab_size = len(vocab)
embedding_dim = 256
rnn_units = 512

model = tf.keras.Sequential([
    tf.keras.layers.Embedding(
        vocab_size,
        embedding_dim
    ),

    tf.keras.layers.SimpleRNN(
        rnn_units,
        return_sequences=True
    ),

    tf.keras.layers.Dense(vocab_size)
])

model.summary()
```

Step 12 — Test the model shape before training

```python
for input_example, target_example in dataset.take(1):
    predictions = model(input_example)

    print("Input shape:", input_example.shape)
    print("Prediction shape:", predictions.shape)
    print("Target shape:", target_example.shape)
```

Step 13 — Create the loss function

```python
loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(
    from_logits=True
)
```

Step 14 — Compile the model

```python
model.compile(
    optimizer='adam',
    loss=loss_fn
)

print("Model compiled successfully!")
```

```python
history = model.fit(
    dataset,
    epochs=5
)
```

```python
prediction_model = tf.keras.Sequential([
    tf.keras.layers.Embedding(vocab_size, embedding_dim),
    tf.keras.layers.SimpleRNN(
        rnn_units,
        return_sequences=True
    ),
    tf.keras.layers.Dense(vocab_size)
])

prediction_model.build((1, 1))

prediction_model.load_weights("/content/rnn_weights.weights.h5")

print("Prediction model ready!")
```

```python
int_to_char = np.array(vocab)

print("Number of characters:", vocab_size)
print("Example mapping:")
print(char_to_int['R'], "→", int_to_char[char_to_int['R']])
```

```python
start_string = "ROMEO: "

input_eval = [char_to_int[s] for s in start_string]

num_generate = 300
text_generated = []

for i in range(num_generate):

    # Keep the latest 100 characters
    input_sequence = input_eval[-100:]

    input_sequence = tf.expand_dims(input_sequence, 0)

    predictions = prediction_model(input_sequence)

    # Take prediction for the final character position
    predictions = predictions[:, -1, :]

    # Randomly select next character
    predicted_id = tf.random.categorical(
        predictions,
        num_samples=1
    )[0, 0].numpy()

    input_eval.append(predicted_id)
    text_generated.append(int_to_char[predicted_id])

print(start_string + ''.join(text_generated))
```
