```python
tokenizer = Tokenizer()

tokenizer.fit_on_texts(df["text"])

word_index = tokenizer.word_index
total_words = len(word_index) + 1

print("Word Index:")
print(word_index)

print("\nTotal words:", total_words)
```

```python
WINDOW_SIZE = 3

print("Window size:", WINDOW_SIZE)
```

Cell 5 — Create input and target pairs

```python
X = []
y = []

for sentence in df["text"]:
    words = sentence.lower().split()

    for i in range(len(words) - WINDOW_SIZE):
        input_words = words[i:i + WINDOW_SIZE]
        target_word = words[i + WINDOW_SIZE]

        X.append(input_words)
        y.append(target_word)

print("Number of samples:", len(X))

print("\nFirst 10 input-target pairs:")

for i in range(10):
    print(X[i], "->", y[i])
```

Cell 6 — Convert the words to numbers

```python
X_numbers = []

for sequence in X:
    X_numbers.append([word_index[word] for word in sequence])

y_numbers = [word_index[word] for word in y]

X_numbers = np.array(X_numbers)
y_numbers = np.array(y_numbers)

print("X shape:", X_numbers.shape)
print("y shape:", y_numbers.shape)

print("\nFirst 10 X:")
print(X_numbers[:10])

print("\nFirst 10 y:")
print(y_numbers[:10])
```

```python
print("X shape:", X_numbers.shape)
print("y shape:", y_numbers.shape)

print("\nOne training example:")
print("Input:", X_numbers[0])
print("Target:", y_numbers[0])
```

```python
model = Sequential()

model.add(Embedding(
    input_dim=total_words,
    output_dim=64,
    input_length=WINDOW_SIZE
))

model.add(SimpleRNN(
    128,
    activation='tanh'
))

model.add(Dense(
    total_words,
    activation='softmax'
))

model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Model compiled successfully.")
```

```python
history = model.fit(
    X_numbers,
    y_numbers,
    epochs=100,
    verbose=0
)

print("Training complete.")
```

```python
def predict_next_word(seed_text):
    words = seed_text.lower().split()

    # Take the last 3 words
    words = words[-WINDOW_SIZE:]

    # Convert words to numbers
    sequence = [word_index[word] for word in words]

    # Add batch dimension
    sequence = np.array([sequence])

    # Predict
    prediction = model.predict(sequence, verbose=0)

    # Get word with highest probability
    predicted_index = np.argmax(prediction[0])

    # Convert number back to word
    for word, index in word_index.items():
        if index == predicted_index:
            return word
```

```python
seed_text = "I love"

predicted_word = predict_next_word(seed_text)

print("Input:", seed_text)
print("Predicted next word:", predicted_word)
```

Cell 13 — Generate multiple words

```python
def generate_text(seed_text, num_words):
    generated_text = seed_text

    for _ in range(num_words):
        next_word = predict_next_word(generated_text)
        generated_text += " " + next_word

    return generated_text
```

```python
def predict_next_word(seed_text):
    words = seed_text.lower().split()

    # Take the last WINDOW_SIZE words
    words = words[-WINDOW_SIZE:]

    # Convert words to numbers
    sequence = [word_index[word] for word in words]

    # Add batch dimension
    sequence = np.array([sequence])

    # Predict probabilities
    prediction = model.predict(sequence, verbose=0)[0]

    # Do not choose the same word as the last word
    last_word = words[-1]
    last_word_index = word_index[last_word]
    prediction[last_word_index] = 0

    # Convert probabilities into a proper distribution
    prediction = prediction / np.sum(prediction)

    # Randomly select based on probabilities
    predicted_index = np.random.choice(
        len(prediction),
        p=prediction
    )

    # Convert number back to word
    for word, index in word_index.items():
        if index == predicted_index:
            return word
```
